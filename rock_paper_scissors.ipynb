```python
from random import randint

# The game funciton loops until user inputs 'X'
# Funciton generates randon comuputer move, followed by user input.
# User and Computer moves are compared as per game rules.

def game():

    moves = ['ROCK', 'PAPER', 'SCISSOR']
    
    while True:  
        
        computer_turn = moves[randint(0, 2)]
        
        my_turn = input("Enter your move (or 'X' to quit): ").upper()

        # For exiting the game
        if my_turn == 'X':
            print("Game Over. Thanks for playing!")
            break
            
        # Handling input erros
        if my_turn not in moves:
            print("Try Again.")
            continue         

        print(f"My Move        : {my_turn}")
        print(f"Computer Move  : {computer_turn}")
        
        if computer_turn == my_turn:
            print("It's a tie!")

        # Game rules
        elif (computer_turn == 'ROCK' and my_turn == 'PAPER') or \
             (computer_turn == 'PAPER' and my_turn == 'SCISSOR') or \
             (computer_turn == 'SCISSOR' and my_turn == 'ROCK'):
            print("You win!")
            
        else:
            print("Computer wins!")
            
game()
```
